## Step 7: Numerical vs Categorical Variables

| Type | Definition | Examples |
|------|------------|----------|
| Numerical - Continuous | Any value in a range | CGPA, AttendancePercent |
| Numerical - Discrete | Countable whole values | Age, Year, Scores |
| Categorical - Nominal | No natural order | Gender, City, Department |
| Categorical - Ordinal | Has natural order | Grade (A>B>C>D>F) |

```python
numerical_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_cols = df.select_dtypes(include=["object"]).columns.tolist()

print(f"NUMERICAL Columns ({len(numerical_cols)}):")
for col in numerical_cols:
    kind = "Discrete" if df[col].nunique() <= 15 else "Continuous"
    print(f"  {col:<22} -> {kind} (unique: {df[col].nunique()})")

print("")
print(f"CATEGORICAL Columns ({len(categorical_cols)}):")
for col in categorical_cols:
    print(f"  {col:<22} -> (unique: {df[col].nunique()})")
```

## Step 8: Missing Values Analysis
Missing values must be identified before any ML work.

```python
missing_count = df.isnull().sum()
missing_pct = (missing_count / len(df) * 100).round(2)
missing_report = pd.DataFrame({"Missing Count": missing_count, "Missing %": missing_pct})
missing_report = missing_report[missing_report["Missing Count"] > 0].sort_values("Missing Count", ascending=False)

print("Missing Values Report:")
print(missing_report.to_string())
print(f"Total missing: {missing_count.sum()} cells")
```

## Step 9: Unique Values in Categorical Columns

```python
for col in categorical_cols:
    if col != "Name":
        vc = df[col].value_counts()
        print(f"Column: {col} | Unique: {df[col].nunique()}")
        print(f"Values: {list(df[col].dropna().unique())}")
        print(vc.to_string())
        print("")
```

## Interview Questions & Answers

**Q1: Structured vs Unstructured data?**  
- Structured: organized in rows/columns (CSV, SQL tables) - like our student dataset  
- Unstructured: no fixed format (text, images, audio) - like student feedback essays

**Q2: Numerical vs Categorical?**  
- Numerical: measurable quantities - Age (22), CGPA (8.5), Score (75)  
- Categorical: labels/groups - Gender ("Male"), Grade ("A"), Department ("CS")

**Q3: Continuous vs Discrete?**  
- Continuous: infinite values in a range - CGPA (7.23...), AttendancePercent  
- Discrete: countable whole values - Year (1,2,3,4), Age (17,18...24)
